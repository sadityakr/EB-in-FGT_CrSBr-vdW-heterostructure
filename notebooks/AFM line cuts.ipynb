```python
import sys as _sys
from pathlib import Path as _Path
try:
    _sys.path.insert(0, str(_Path(__vsc_ipynb_file__).parent.parent))
except NameError:
    pass  # Not in VS Code; rely on pip install -e .

import os
from scripts.utils import setup_notebook, OKABE_ITO_CYCLE

PROJECT_ROOT, np, pd, plt, Path = setup_notebook()

# Backward-compatible alias
color_values = OKABE_ITO_CYCLE
```

```python
RAW_DATA_DIR = PROJECT_ROOT / "data" / "AFM"
PLOTS_DIR    = PROJECT_ROOT / "figures"
plot_name    = PLOTS_DIR / "AFM"
PLOTS_DIR.mkdir(exist_ok=True)
```

```python
# Device 1


data = pd.read_csv(os.path.join(RAW_DATA_DIR,r"crsbr_fgt_device1_20250430_cut_data"), skiprows=3, delimiter=r"  ", names = ["x", "h"])

plt.plot(data["x"]*1e6, data["h"]*1e9)
plt.xlabel("x (µm)")
plt.ylabel("Height (nm)")
plt.yticks(np.arange(40, 190, 20))



plt.savefig(os.path.join(PLOTS_DIR,"Device 1 cut.png"), dpi=300, bbox_inches='tight')
```
