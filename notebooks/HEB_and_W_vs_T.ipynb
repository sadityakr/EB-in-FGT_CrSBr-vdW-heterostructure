```python
import sys as _sys
from pathlib import Path as _Path
try:
    _sys.path.insert(0, str(_Path(__vsc_ipynb_file__).parent.parent))
except NameError:
    pass  # Not in VS Code; rely on pip install -e .

import os
from scripts.utils import setup_notebook, OKABE_ITO_CYCLE

PROJECT_ROOT, np, pd, plt, Path = setup_notebook()

# Backward-compatible alias
color_values = OKABE_ITO_CYCLE
```

```python
RESULTS_DIR = PROJECT_ROOT / "results"
PLOTS_DIR   = PROJECT_ROOT / "figures"
PLOTS_DIR.mkdir(exist_ok=True)
```

```python
df1 = pd.read_csv("../results/PFP2500mT_Tvs_EB.csv")
df2 = pd.read_csv("../results/PFPM2500mT_Tvs_EB.csv")
```

```python
# === Plot ===
fig, ax = plt.subplots(figsize=(5, 4), dpi=600)

ax.errorbar(df1["T"], df1["HEB"], yerr=5,
            fmt='o-', label="PF: 2.5 T", color=color_values[0], capsize=3)

ax.errorbar(df2["T"], df2["HEB"], yerr=5,
            fmt='s-', label="PF: -2.5 T", color=color_values[1], capsize=3)

ax.set_xlabel("$T$ $(K)$")
ax.set_ylabel("$H_{EB}$ $(mT)$")
ax.legend(frameon=False)
plt.tight_layout()
fig.savefig(PLOTS_DIR / "EBvsT.png")
plt.show()
```

```python
# === Plot ===
fig, ax = plt.subplots(figsize=(5, 4), dpi=600)

ax.errorbar(df1["T"], df1["W"], yerr=5,
            fmt='o-', label="PF: 2.5 T", color=color_values[0], capsize=3)

ax.errorbar(df2["T"], df2["W"], yerr=5,
            fmt='s-', label="PF: -2.5 T", color=color_values[1], capsize=3)

ax.set_xlabel("$T$ $(K)$")
ax.set_ylabel("$H_{C}$ $(mT)$")
ax.legend(frameon=False)
plt.tight_layout()
fig.savefig(PLOTS_DIR / "WvsT.png")
plt.show()
```

```python
# === Plot ===
fig, ax = plt.subplots(figsize=(5, 4), dpi=600)

ax.errorbar(df1["T"], df1["C1"], yerr=5,
            fmt='o-', label="$H_{C}^{-}$", color=color_values[0], capsize=3)

ax.errorbar(df1["T"], df1["C2"], yerr=5,
            fmt='s-', label="$H_{C}^{+}$", color=color_values[3], capsize=3)

ax.set_xlabel("$T$ $(K)$")
ax.set_ylabel("$H_{C}$ $(mT)$")
ax.legend(frameon=False)
plt.tight_layout()
fig.savefig(PLOTS_DIR / "C1_C2_vsT.png")
plt.show()
```
